### import necessary libraries

```python
import pandas as pd
from sklearn.datasets import fetch_california_housing
from sklearn.linear_model import LinearRegression, Ridge, ElasticNet, Lasso
from matplotlib import pyplot as plt
from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings("ignore")
```

### loaded the inbuilt dataset from sklearn library

```python
data = fetch_california_housing()
```

```python
data
```

```python
fetch_california_housing_data = pd.DataFrame(data = data.data,columns=data.feature_names)
fetch_california_housing_data["Target"] = data.target
fetch_california_housing_data.head()
```

### Model Building

```python
X = fetch_california_housing_data.drop(labels="Target",axis=1)
y = fetch_california_housing_data["Target"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=101)
```

```python
X_train.shape,y_train.shape
```

```python
X_test.shape,y_test.shape
```

```python
Linear_model = LinearRegression()
Ridge_model = Ridge()
Lasso_model = Lasso()
ElasticNet_model = ElasticNet()
```

### Model Training

```python
Linear_model.fit(X_train,y_train)
```

```python
Ridge_model.fit(X_train,y_train)
```

```python
Lasso_model.fit(X_train,y_train) 
```

```python
ElasticNet_model.fit(X_train,y_train)
```

```python
print("Linear Model Coefficients:")
print(Linear_model.coef_)

print("Ridge Model Coefficients:")
print(Ridge_model.coef_)

print("Lasso Model Coefficients:")
print(Lasso_model.coef_)

print("Elastic Net Model Coefficients:")
print(ElasticNet_model.coef_)
```

```python
def plot_coefficients(model,feature_names,title):

    coef_df = pd.DataFrame(
        {
            "Feature":feature_names,
            "Coefficient":model.coef_
        }
    )

    coef_df = coef_df.sort_values(by="Coefficient")
    print(coef_df)

    plt.figure(figsize=(10,6))

    plt.barh(coef_df["Feature"],coef_df["Coefficient"])

    plt.title(title)

    plt.xlabel("Coefficient Value")

    plt.ylabel("Features")

    plt.tight_layout()

    plt.show()
```

```python
plot_coefficients(Linear_model,X.columns,"Linear Regression Coefficients")
```

```python
plot_coefficients(Ridge_model,X.columns,"Ridge Regression Coefficients")
```

```python
plot_coefficients(Lasso_model,X.columns,"Lasso Regression Coefficients")
```

```python
plot_coefficients(ElasticNet_model,X.columns,"Elastic Net Regression Coefficients")
```
