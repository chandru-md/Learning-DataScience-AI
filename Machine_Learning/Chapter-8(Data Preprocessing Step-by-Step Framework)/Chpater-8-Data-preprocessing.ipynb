```python
"""continuous - std scalar,minmax scalar, robust scalar\n
discrete - one hot encoder, label encoder"""
```

```python
weather_data.head()
```

```python
weather_data_copy = weather_data.copy()
```

```python
weather_data_copy.isna().sum()
```

```python
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
```

```python
le_encoder = LabelEncoder()
weather_data_copy['Weather'] = le_encoder.fit_transform(weather_data_copy["Weather"])
```

```python
weather_data_copy.dtypes
```

#### One hot encoder - 
- This can be achieved by using any of the following 2 libraries:

- pandas = pd.get_dummies()

- sklearn = OneHotEncoder()

```python
weather_data_copy_2 = weather_data_copy.copy()
```

```python
weather_data_copy_2 = pd.get_dummies(data =weather_data_copy_2,columns=["Weather"],dtype="int")
```

```python
weather_data_copy_2
```

```python
weather_data_copy_2.dtypes
```

```python
weather_data_copy_3 = weather_data_copy.copy()
```

```python
ohe = OneHotEncoder()
```

```python
weather_data_copy_3 = ohe.fit_transform(weather_data_copy_3[["Weather"]]).toarray()
```
