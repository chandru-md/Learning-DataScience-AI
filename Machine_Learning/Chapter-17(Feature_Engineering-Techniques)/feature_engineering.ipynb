```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer
from sklearn.metrics import accuracy_score,confusion_matrix,precision_score,recall_score
from sklearn.feature_selection import RFE,SelectFromModel
import warnings
warnings.filterwarnings("ignore")
from sklearn.ensemble import RandomForestClassifier
```

Feature Engineering

whatever you do with features we call it as feature engineering.

    Feature Elimination - dropping the features.
    Feature addition - adding some features.
    Feature Transformation - transforming the given feature values into another scale - log transformation, sqrt transformation.
    Feature Selection - deciding which feature are important out of many features and the choosing that features for model building. 

Feature Selection Techniques

    Sklearn - SelectFromModel
    Sklearn - RFE(i.e. Recursive Feature Elimination)


```python
breast_cancer = load_breast_cancer()
```

```python
breast_cancer_data = pd.DataFrame(data=breast_cancer.data,columns=breast_cancer.feature_names)
breast_cancer_data["Target"] = breast_cancer.target
breast_cancer_data.head()
```

```python
# Model Building
```

```python
X = breast_cancer.data
y = breast_cancer.target
```

```python
X
```

```python
y
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=101,shuffle=True,stratify=y)
```

```python
X_train.shape,y_train.shape
```

```python
X_test.shape,y_test.shape
```

```python
len(X_train),len(y_train)
```

```python
len(X_test),len(y_test)
```

```python
# Model Training, Testing and Evaluation
```

```python
rf_model = RandomForestClassifier(max_depth=5)
rf_model.fit(X_train,y_train)
y_pred_test = rf_model.predict(X_test)
print("Accuracy Score:\n",round(accuracy_score(y_test,y_pred_test),4))
print("Precision Score:\n",round(precision_score(y_test,y_pred_test),4))
print("Recall Score:\n",round(recall_score(y_test,y_pred_test),4))
print("Confusion Matrix:\n",confusion_matrix(y_test,y_pred_test))
```
