### MODEL EVAKVALIDATION (TESTs)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split,KFold,cross_val_score,LeaveOneOut
```

```python
claiments_data = pd.read_csv("claiments.csv")
```

```python
claiments_data.head()
```

```python
claiments_data.shape
```

```python
claiments_data.dtypes
```

```python
claiments_data.isna().sum()
```

```python
claiments_data.dropna(inplace=True)
```

```python
claiments_data.isnull().sum()
```

```python
claiments_data.head()
```

```python
X = claiments_data.drop(labels=["CASENUM","ATTORNEY"],axis=1)
```

```python
y = claiments_data["ATTORNEY"]
```

```python
log_model = LogisticRegression()
```

```python
log_model.fit(X,y)
```

```python
y_pred = log_model.predict(X)
```

```python
y_pred
```

### first technique ===== TRAIN TEST SPLIT

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=101,shuffle=True)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

```python
log_model.fit(X_train,y_train)
```

```python
y_pred_train = log_model.predict(X_train)
y_pred_train
```

```python
y_pred_test = log_model.predict(X_test)
y_pred_test
```

```python
accuracy_score(y_train,y_pred_train)
```

```python
accuracy_score(y_test,y_pred_test)
```

### second technique ===== K-FOLD CROSS VALIDATION

```python
#1
```

```python
k_fold =KFold(n_splits=5, shuffle=True, random_state=101)
cv_scores = cross_val_score(estimator=log_model,
    X=X,
    y=y,
    cv=k_fold)
```

```python
print(cv_scores)

print("Overall Accuracy: ",cv_scores.mean().round(2))
print("Overall Standard Deviation: ",cv_scores.std().round(2))
```

```python
#2
```

```python
k_fold =KFold(n_splits=5, shuffle=True, random_state=101)
cv_scores = cross_val_score(estimator=log_model,
    X=X_train,
    y=y_train,
    cv=k_fold)
```

```python
print(cv_scores)

print("Overall Accuracy: ",cv_scores.mean().round(2))
print("Overall Standard Deviation: ",cv_scores.std().round(2))
```

### third technique ===== Leave one out cross validation

```python
#1
```

```python
loocv = LeaveOneOut()

cv_scores = cross_val_score(estimator=log_model,
    X=X,
    y=y,
    cv=loocv)
```

```python
print(cv_scores)

print("Overall Accuracy: ",cv_scores.mean().round(2))
print("Overall Standard Deviation: ",cv_scores.std().round(2))
```

```python
#2
```

```python
loocv = LeaveOneOut()

cv_scores = cross_val_score(estimator=log_model,
    X=X_train,
    y=y_train,
    cv=loocv)
```

```python
print(cv_scores)

print("Overall Accuracy: ",cv_scores.mean().round(2))
print("Overall Standard Deviation: ",cv_scores.std().round(2))
```
