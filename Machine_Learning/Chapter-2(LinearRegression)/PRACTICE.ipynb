#### 1.Import Neccessary Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
from sklearn.linear_model import LinearRegression
```

#### 4. Assumptions to check to perform Linear regression

#### 4.1 Linearity and Normality

```python
sns.lmplot(data=cars_data,x="HP",y="MPG") # with straight line
```

```python
sns.scatterplot(data=cars_data,x="HP",y="MPG") #without straight line
```

```python
for i in cars_data:
    sns.lmplot(data=cars_data,x=i,y="MPG") # with straight line
    plt.show()
```

```python
for i in cars_data:
    sns.distplot(x=cars_data[i])
    plt.show()
```

#### Conclude that both test1 (Normality and Linearity) be failed.

#### 4.2 Multi CoLinearity

```python
cars_data.corr()
```

```python
corr_matrix = cars_data.corr()
```

```python
sns.heatmap(data=corr_matrix,annot=True)
plt.show()
```

#### test 2, there is multicolinearity in this data and failed

#### 4.3 Auto Regression

#### no time dependent data present in this data, Passed

#### 5. Data Preparation

```python
cars_data
```

```python
X = cars_data.drop(labels="MPG",axis=1)
```

```python
y = cars_data['MPG']
```

```python
print(X)
```

```python
print(y)
```

#### 6. Model Building

```python
linear_model = LinearRegression()
```

#### 7. Model Training

```python
linear_model.fit(X,y)
```

```python
linear_model.intercept_
```

```python
linear_model.coef_
```

#### 8. Model Testing

```python
print(y)
```

```python
y_pred = linear_model.predict(X)
y_pred
```

#### 9. Model Evaluation

```python
error = y-y_pred
```

```python
error
```

#### 4.4 Homoscedasticity Test

```python
sns.scatterplot(data=cars_data,x='HP',y=error)
```

```python
sns.scatterplot(data=cars_data,x='SP',y=error)
```

```python
sns.scatterplot(data=cars_data,x='WT',y=error)
```

```python
sns.scatterplot(data=cars_data,x='VOL',y=error)
```

#### 4.5 Zero Residual Mean TEST

```python
sns.scatterplot(x=y,y=y_pred)
```

#### finally homoscedasticity and zero residual mean test - failed
