### Import necessary libraries

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report,precision_score,recall_score
from sklearn.ensemble import AdaBoostClassifier,GradientBoostingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
import warnings
warnings.filterwarnings("ignore")
```

### Data Preparation
#### 1. data cleaning
#### 2. data transformation

```python
del credit_card_data['ID']
```

### Label encoder processing

```python
credit_card_data.head()
```

```python
le = LabelEncoder()
credit_card_data["SEX"] = le.fit_transform(credit_card_data["SEX"])
credit_card_data["EDUCATION"] = le.fit_transform(credit_card_data["EDUCATION"])
credit_card_data["MARRIAGE"] = le.fit_transform(credit_card_data["MARRIAGE"])
```

```python
credit_card_data.head()
```

```python
X = credit_card_data.drop("DEFAULT",axis=1)

y = credit_card_data['DEFAULT']
```

```python
## Model building
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size=0.80,shuffle=True,random_state=101,stratify=y)
```

```python
X.shape,y.shape
```

```python
X_train.shape
```

```python
X_test.shape
```

### Model Training

### Training time analysis

```python
adab_classifier = AdaBoostClassifier()
gdb_classifier = GradientBoostingClassifier(n_estimators=50)
xgb_classifier = XGBClassifier(n_estimators=50)
lgb_classifier = LGBMClassifier(n_estimators=50)
```

```python
%%time
adab_classifier.fit(X_train,y_train)
```

```python
%%time
gdb_classifier.fit(X_train,y_train)
```

```python
%%time
xgb_classifier.fit(X_train,y_train)
```

```python
%%time
lgb_classifier.fit(X_train,y_train)
```

### Model_prediction

### 7. Model testing

```python
y_pred_adab = adab_classifier.predict(X_test)
y_pred_gdb = gdb_classifier.predict(X_test)
y_pred_xgb = xgb_classifier.predict(X_test)
y_pred_lgb = lgb_classifier.predict(X_test)
```

### Model evaluation

### Ada-boost - evaluation

```python
print("Accuracy-Score:",round(accuracy_score(y_test,y_pred_adab),4))
print("Precision-Score",round(precision_score(y_test,y_pred_adab),4))
print("Recall-Score:",round(recall_score(y_test,y_pred_adab),4))
print("Confusion-Matrix:\n",confusion_matrix(y_test,y_pred_adab))
```

### gradient-boosting evaluation

```python
print("Accuracy-Score:",round(accuracy_score(y_test,y_pred_gdb),4))
print("Precision-Score",round(precision_score(y_test,y_pred_gdb),4))
print("Recall-Score:",round(recall_score(y_test,y_pred_gdb),4))
print("Confusion-Matrix:\n",confusion_matrix(y_test,y_pred_gdb))
```

### Xgboost-evaluation

```python
print("Accuracy-Score:",round(accuracy_score(y_test,y_pred_xgb),4))
print("Precision-Score",round(precision_score(y_test,y_pred_xgb),4))
print("Recall-Score:",round(recall_score(y_test,y_pred_xgb),4))
print("Confusion-Matrix:\n",confusion_matrix(y_test,y_pred_xgb))
```

### Light Gradient-boosting evaluation

```python
print("Accuracy-Score:",round(accuracy_score(y_test,y_pred_lgb),4))
print("Precision-Score",round(precision_score(y_test,y_pred_lgb),4))
print("Recall-Score:",round(recall_score(y_test,y_pred_lgb),4))
print("Confusion-Matrix:\n",confusion_matrix(y_test,y_pred_lgb))
```
