```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix
from sklearn.ensemble import AdaBoostClassifier,GradientBoostingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.tree import DecisionTreeClassifier
import warnings
warnings.filterwarnings("ignore")
```

```python
credit_card_data = pd.read_csv("credit_card_clean.csv")
```

```python
credit_card_data.head()
```

```python
credit_card_data.shape
```

```python
credit_card_data.dtypes
```

```python
credit_card_data.isna().sum()
```

```python
le = LabelEncoder()
```

```python
credit_card_data["SEX"] = le.fit_transform(credit_card_data["SEX"])
credit_card_data["EDUCATION"] = le.fit_transform(credit_card_data["EDUCATION"])
credit_card_data["MARRIAGE"] = le.fit_transform(credit_card_data["MARRIAGE"])
```

```python
credit_card_data.head()
```

```python
X = credit_card_data.drop(labels="DEFAULT",axis=1)
y = credit_card_data["DEFAULT"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,shuffle=True,random_state=101,stratify=y)
```

```python
adaboost_classifier = AdaBoostClassifier()

adaboost_classifier.fit(X_train,y_train)
```

```python
y_pred_adaboost = adaboost_classifier.predict(X_test)
```

```python
print("Accuracy Score:",round(accuracy_score(y_test,y_pred_adaboost),4))
print("Precision Score:",round(precision_score(y_test,y_pred_adaboost),4))
print("Recall Score:",round(recall_score(y_test,y_pred_adaboost),4))
print("Confusion Matrix:\n",confusion_matrix(y_test,y_pred_adaboost))
```

```python
gradient_boost_classifier = GradientBoostingClassifier()

gradient_boost_classifier.fit(X_train,y_train)
```

```python
y_pred_grad_boost = gradient_boost_classifier.predict(X_test)
```

```python
print("Accuracy Score:",round(accuracy_score(y_test,y_pred_grad_boost),4))
print("Precision Score:",round(precision_score(y_test,y_pred_grad_boost),4))
print("Recall Score:",round(recall_score(y_test,y_pred_grad_boost),4))
print("Confusion Matrix:\n",confusion_matrix(y_test,y_pred_grad_boost))
```

```python
Extreme_grad_boost = XGBClassifier()

Extreme_grad_boost.fit(X_train,y_train)
```

```python
y_pred_XGB = Extreme_grad_boost.predict(X_test)
```

```python
print("Accuracy Score:",round(accuracy_score(y_test,y_pred_XGB),4))
print("Precision Score:",round(precision_score(y_test,y_pred_XGB),4))
print("Recall Score:",round(recall_score(y_test,y_pred_XGB),4))
print("Confusion Matrix:\n",confusion_matrix(y_test,y_pred_XGB))
```

```python
light_grad_boost = LGBMClassifier()

light_grad_boost.fit(X_train,y_train)
```

```python
y_pred_lgbm = light_grad_boost.predict(X_test)
```

```python
print("Accuracy Score:",round(accuracy_score(y_test,y_pred_lgbm),4))
print("Precision Score:",round(precision_score(y_test,y_pred_lgbm),4))
print("Recall Score:",round(recall_score(y_test,y_pred_lgbm),4))
print("Confusion Matrix:\n",confusion_matrix(y_test,y_pred_lgbm))
```
