# Classifier

### NECESSARY LIBRARIES

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,recall_score,precision_score,confusion_matrix
from sklearn.ensemble import GradientBoostingClassifier
from datetime import datetime
import xgboost as xgb
```

```python
breast_cancer_data = load_breast_cancer()
```

```python
breast_cancer_dataset = pd.DataFrame(data = breast_cancer_data.data,columns=breast_cancer_data.feature_names)
breast_cancer_dataset['Target'] = breast_cancer_data.target
```

```python
breast_cancer_dataset
```

```python
breast_cancer_dataset.isna().sum()
```

```python
breast_cancer_dataset.shape
```

```python
breast_cancer_dataset.dtypes
```

```python
X = breast_cancer_dataset.drop(labels="Target",axis=1)
```

```python
y = breast_cancer_dataset["Target"]
```

```python
std_scaler = StandardScaler()
```

```python
scaled_X = std_scaler.fit_transform(X)
```

```python
y
```

```python
gb_classifier = GradientBoostingClassifier(n_estimators=50,max_depth=3)
```

```python
X_train,X_test,y_train,y_test = train_test_split(scaled_X,y,test_size=0.20,shuffle=True, random_state=101,stratify=y)
```

```python
start_time = datetime.now()
gb_classifier.fit(X_train,y_train)
end_time = datetime.now()

print("Time taken to train the algorithm:",end_time-start_time)
```

### MODEL EVALUATION

#### TRAIN DATA

```python
y_pred_train = gb_classifier.predict(X_train)
print("Accuracy Score: ",accuracy_score(y_train,y_pred_train))
print("Recall Score: ",recall_score(y_train,y_pred_train))
print("Precision Score: ",precision_score(y_train,y_pred_train))
print("Confusion Matrix:\n ",confusion_matrix(y_train,y_pred_train))
```

#### TEST DATA

```python
y_pred_test = gb_classifier.predict(X_test)
print("Accuracy Score: ",accuracy_score(y_test,y_pred_test))
print("Recall Score: ",recall_score(y_test,y_pred_test))
print("Precision Score: ",precision_score(y_test,y_pred_test))
print("Confusion Matrix:\n ",confusion_matrix(y_test,y_pred_test))
```
